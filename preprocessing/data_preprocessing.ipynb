```python
import re
from datasets import Dataset, load_dataset
from sentence_transformers import SentenceTransformer
import Stemmer
import bm25s
```

```python
# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("common-pile/caselaw_access_project", split="train[:15000]")
print(ds)
```

```python
ds = ds.shuffle(seed=42)

train_test = ds.train_test_split(test_size=0.2)
test_val = train_test["test"].train_test_split(test_size=0.5)

train_ds = train_test["train"]
val_ds = test_val["train"]
test_ds = test_val["test"]
```

```python
print(train_ds.column_names)
print(train_ds[0])
```

```python
def flatten(example):
    return {
        "case_id": example["id"],
        "text": example["text"],
        "source": example["source"],
        "author": example["metadata"].get("author"),
        "url": example["metadata"].get("url"),
    }

train_ds = train_ds.map(flatten)
```

```python
def clean_text(example):
    text = example["text"]

    # remove excessive newlines
    text = re.sub(r"\n+", "\n", text)

    # normalize spaces
    text = re.sub(r"[ \t]+", " ", text)

    # strip leading/trailing
    text = text.strip()

    return {"clean_text": text}

train_ds = train_ds.map(clean_text)
```

```python
def split_paragraphs(example):
    paragraphs = example["clean_text"].split("\n")
    paragraphs = [p.strip() for p in paragraphs if len(p.strip()) > 50]
    return {"paragraphs": paragraphs}

train_ds = train_ds.map(split_paragraphs)
```

```python
def create_chunks(example, max_words=200, overlap=50):
    chunks = []
    current_chunk = []
    current_len = 0

    for para in example["paragraphs"]:
        words = para.split()
        
        if current_len + len(words) > max_words:
            chunks.append(" ".join(current_chunk))
            current_chunk = current_chunk[-overlap:]  # overlap
            current_len = len(current_chunk)

        current_chunk.extend(words)
        current_len += len(words)

    if current_chunk:
        chunks.append(" ".join(current_chunk))

    return {"chunks": chunks}

train_ds = train_ds.map(create_chunks)
```

```python
print(train_ds.column_names)
print(train_ds[0])
```

```python
train_chunks_list = []
for example in train_ds:
    for chunk in example["chunks"]:
        train_chunks_list.append({
            "text": chunk,
            "case_id": example["id"],
            "source": example["source"],
            "author": example["metadata"].get("author"),
            "url": example["metadata"].get("url"),
        })

train_chunks = Dataset.from_list(train_chunks_list)
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
def embed(batch):
    embeddings = model.encode(batch["text"], batch_size=32)
    return {"embedding": embeddings.tolist()}

train_chunks = train_chunks.map(embed, batched=True)
```

```python
train_chunks.to_parquet("embeddings.parquet")
```

```python
from dataclasses import dataclass

@dataclass
class Chunk:
    text: str
    case_id: str
    chunk_idx: int
    source: str = ""
    author: str = ""
    url: str = ""
```

```python
chunks = [
    Chunk(
        text=row["text"],
        case_id=row["case_id"],
        chunk_idx=i,
        source=row.get("source", ""),
        author=row.get("author", ""),
        url=row.get("url", ""),
    )
    for i, row in enumerate(train_chunks_list)
]
```

```python
import json
from pathlib import Path

def build_bm25(chunks, save_dir="bm25_index"):
    corpus = [c.text for c in chunks]
    tokens = bm25s.tokenize(corpus, stopwords="en")
    retriever = bm25s.BM25()
    retriever.index(tokens)
    retriever.save(save_dir)

    Path(save_dir).mkdir(parents=True, exist_ok=True)
    with open(f"{save_dir}/chunks_meta.json", "w") as f:
        json.dump([{"text": c.text, "case_id": c.case_id, "chunk_idx": c.chunk_idx, "source": c.source, "author": c.author, "url": c.url} for c in chunks], f)

    return retriever, chunks

build_bm25(chunks, save_dir="chunks")
```
